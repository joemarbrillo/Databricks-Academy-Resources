### Task 1: View the version history

Use `DESCRIBE HISTORY` to see the full change log for `new_employees`.

```python
%sql
-- TODO: Show the version history of new_employees
DESCRIBE HISTORY new_employees;
```

<details>
<summary>Hint</summary>

<pre><code>DESCRIBE HISTORY new_employees;</code></pre>

</details>

Look at the **operation** column for each version. Can you match each version to what you did in Lab 05?
- Version 0 should be `CREATE TABLE`
- Version 1 should be `WRITE` (the INSERT)
- Version 2 should be `UPDATE`
- Version 3 should be `OPTIMIZE`
- Version 4 should be `DELETE`
- Version 5 should be `OPTIMIZE`

### Task 2: Query the original version

Query `new_employees` as it was at **version 0**, right after it was created from `employees2.csv`, before any modifications.

```python
%sql
-- TODO: Query new_employees at version 0
SELECT * FROM new_employees VERSION AS OF 0;
```

<details>
<summary>Hint</summary>

<pre><code>SELECT * FROM new_employees VERSION AS OF 0;</code></pre>

</details>

You should see the original 2 rows: Maria as `Data Engineer` and Aiden as `Data Analyst`. No Priya, and no role changes yet.

### Task 3: Find a specific change

In Lab 05, you updated Aiden's role. Query the version **right after the INSERT but before the UPDATE** to see Aiden's original role alongside Priya's row.

```python
%sql
-- TODO: Query the version that has all 3 rows with original roles (after INSERT, before UPDATE)
SELECT * FROM new_employees VERSION AS OF 1;
```

<details>
<summary>Hint</summary>

The INSERT was version 1. Query that version:
<pre><code>SELECT * FROM new_employees VERSION AS OF 1;</code></pre>

</details>

You should see 3 rows: Maria, Aiden (still `Data Analyst`), and Priya. This is the state between the INSERT and the UPDATE.

### Task 4: Compare current vs. original

Write a query that shows the row count for both the **current version** and **version 0** of `new_employees`. Use UNION ALL to combine the results.

```python
%sql
-- TODO: Compare row counts between current and version 0
SELECT 'Current' AS version, COUNT(*) as row_count FROM new_employees
UNION ALL
SELECT 'Version 0' AS version, COUNT(*) as row_count FROM new_employees VERSION AS OF 0;
```

<details>
<summary>Hint</summary>

<pre><code>SELECT 'Current' AS version, COUNT(*) AS row_count FROM new_employees
UNION ALL
SELECT 'Version 0', COUNT(*) FROM new_employees VERSION AS OF 0;</code></pre>

</details>

Both should show **2 rows**. You started with 2, added 1, then deleted 1. The data is the same size, but the content is different (Aiden's role changed). This is why comparing row counts alone isn't enough; sometimes you need to compare the actual data.
