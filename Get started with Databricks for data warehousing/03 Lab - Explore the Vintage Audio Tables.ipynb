### Task 4: Count stores by region - spot the correct query

Vintage Audio Co. operates 30 stores grouped into four US regions: `Northeast`, `Southeast`, `Midwest`, and `West`. You want **one row per region with the number of stores in it**.

Below are **two** queries. Both run without error, but only one answers the question. **Run them both**, look at the results, and decide which one is correct and why.


```python
%sql
-- Query A
SELECT
  region,
  COUNT(*) AS store_count
FROM stores
GROUP BY region
ORDER BY store_count DESC;
```

```python
%sql
-- Query B
SELECT
  region,
  state,
  COUNT(*) AS store_count
FROM stores
GROUP BY region, state
ORDER BY store_count DESC;
```

<details>
<summary><strong>Which one is correct?</strong></summary>
<div style="margin-top:8px; font-size:14px;">
<p><strong>Query A is correct.</strong> Grouping by <code>region</code> alone gives exactly one row per region, which is what the question asks for.</p>
<p><strong>Query B</strong> groups by <code>region</code> <em>and</em> <code>state</code>, so it returns one row per region-and-state combination &mdash; many more than four rows, splitting each region's count across its states. It answers a different question (&ldquo;stores per state within each region&rdquo;).</p>
<p>The lesson: the columns you <code>GROUP BY</code> decide the grain of the result. Add only the columns you actually want one row per.</p>
</div>
</details>


Query A returns four rows. The **West** region has the most stores (10), followed by **Midwest** and **Southeast** (7 each), then **Northeast** (6) - a national footprint that's reasonably well-distributed.
