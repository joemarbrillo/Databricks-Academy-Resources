```python
# Clear all tables in the `retail` schema.
# NOTE: DROP TABLE permanently removes the table and its data.
# If you instead want to keep the tables but remove their rows,
# replace `DROP TABLE` with `TRUNCATE TABLE` below.

catalog = spark.sql("SELECT current_catalog() AS catalog").collect()[0]["catalog"]
schema = "retail"

tables = (
    spark.sql(f"SHOW TABLES IN `{catalog}`.`{schema}`")
    .filter("isTemporary = false")
    .collect()
)

if not tables:
    print(f"No tables found in {catalog}.{schema}.")

for t in tables:
    spark.sql(f"DROP TABLE IF EXISTS `{catalog}`.`{schema}`.`{t.tableName}`")
    print(f"Dropped {catalog}.{schema}.{t.tableName}")
```

```python
# Clear all volumes (and tables) in the `retail` schema, then drop the catalog.
# NOTE: DROP permanently removes the objects and their data.

catalog = "learn_get_started_da_joemar_brillo"
schema = "retail"

# Drop all volumes in the schema.
volumes = (
    spark.sql(f"SHOW VOLUMES IN `{catalog}`.`{schema}`")
    .collect()
)

if not volumes:
    print(f"No volumes found in {catalog}.{schema}.")

for v in volumes:
    spark.sql(f"DROP VOLUME IF EXISTS `{catalog}`.`{schema}`.`{v.volumeName}`")
    print(f"Dropped volume {catalog}.{schema}.{v.volumeName}")

# Drop all tables in the schema.
tables = (
    spark.sql(f"SHOW TABLES IN `{catalog}`.`{schema}`")
    .filter("isTemporary = false")
    .collect()
)

if not tables:
    print(f"No tables found in {catalog}.{schema}.")

for t in tables:
    spark.sql(f"DROP TABLE IF EXISTS `{catalog}`.`{schema}`.`{t.tableName}`")
    print(f"Dropped table {catalog}.{schema}.{t.tableName}")

# Finally, delete the catalog.
spark.sql(f"DROP CATALOG IF EXISTS `{catalog}` CASCADE")
print(f"Dropped catalog {catalog}")
```
