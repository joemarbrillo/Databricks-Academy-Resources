### Task 6: Recover from a destructive mistake

A teammate runs the cell below by accident. It clears the entire shelf. Run it as-is to set up the recovery scenario, then use `RESTORE TABLE` to put the books back.

```sql
%sql
-- Run as-is. This is the "mistake" you'll recover from.
DELETE FROM book_inventory;
SELECT COUNT(*) AS rows_remaining FROM book_inventory;
```

Zero rows. Now look at the version history again. You need the version *right before* the `DELETE` so the `RESTORE` puts the four books back with the correct `in_stock` for *Clean Code*.

```sql
%sql
DESCRIBE HISTORY book_inventory;
```

In the history above, find the row whose `operation` is `UPDATE` (the correction from Task 3). That's the version you want to restore to. Fill in its version number below.

```sql
%sql
-- TODO: Restore book_inventory to the version just BEFORE the DELETE
RESTORE TABLE book_inventory VERSION AS OF 3;
```
